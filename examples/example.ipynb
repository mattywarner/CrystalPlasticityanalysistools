```python
import CPanalysistools
from CPanalysistools import *
```

```python
material_IDs = np.load("./data/polycrystal_128_VE_material_IDs.npy")
```

```python
pheno_strain = np.load("./data/pheno_128_xx_strain.npy")
dislo_strain = np.load("./data/dislo_128_xx_strain.npy")
```

```python
pheno_triax = np.load("./data/pheno_triax.npy")
dislo_triax = np.load("./data/dislo_triax.npy")
```

```python
max_pheno_strain_from_grains = CPanalysistools.max_val_from_grains(pheno_strain, material_IDs, 1000)
max_dislo_strain_from_grains = CPanalysistools.max_val_from_grains(dislo_strain, material_IDs, 1000)
```

```python
max_pheno_triax_from_grains = CPanalysistools.max_val_from_grains(pheno_triax, material_IDs, 1000)
max_dislo_triax_from_grains = CPanalysistools.max_val_from_grains(dislo_triax, material_IDs, 1000)
```

```python
%matplotlib inline
CPanalysistools.comparison_plot(max_pheno_strain_from_grains, 
                                max_dislo_strain_from_grains, 
                                r" Maximum grain strains, $\varepsilon_{xx}^{\text{pheno}}$", 
                                r" Maximum grain strains, $\varepsilon_{xx}^{\text{dislo}}$")
```

```python
%matplotlib inline
CPanalysistools.comparison_plot(max_pheno_triax_from_grains, 
                                max_dislo_triax_from_grains, 
                                r" Maximum grain stress triaxiality, $T_{\sigma}^{\text{pheno}}$", 
                                r" Maximum grain stress triaxiality, $T_{\sigma}^{\text{dislo}}$")
```
